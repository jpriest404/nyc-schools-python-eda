```python
import pandas as pd

# Read in the data
schools = pd.read_csv("schools.csv")

# Preview the data
schools.head()
```

# 1. Finding schools with the best math scores
# Subset the data to find the schools with math scores of at least 80% (640+ out of 800).
# Save as best_math_schools containing "school_name" and "average_math", sorted descending.

```python
best_math_schools = schools.loc[schools['average_math'] >= 640]
best_math_schools = (
    best_math_schools[['school_name', 'average_math']]
    .sort_values(by='average_math', ascending=False)
    .reset_index(drop=True)
)
best_math_schools
```

# 2. Identifying the top 10 performing schools
# Find the 10 best performing schools based on total SAT score across three sections.
# Save as top_10_schools containing "school_name" and "total_SAT".

```python
schools['total_SAT'] = schools['average_math'] + schools['average_reading'] + schools['average_writing']

top_10_schools = (
    schools[['school_name', 'total_SAT']]
    .sort_values(by='total_SAT', ascending=False)
    .head(10)
    .reset_index(drop=True)
)
top_10_schools
```

# 3. Locating the NYC borough with the largest standard deviation in SAT performance
# Find num_schools, average_SAT, and std_SAT per borough, rounded to two decimal places.

```python
schools['total_SAT'] = schools['average_math'] + schools['average_reading'] + schools['average_writing']

largest_std_dev = (
    schools
    .groupby('borough')
    .agg(
        num_schools=('school_name', 'count'),
        average_SAT=('total_SAT', 'mean'),
        std_SAT=('total_SAT', 'std')
    )
    .reset_index()
    .sort_values(by='std_SAT', ascending=False)
    .head(1)
    .round(2)
)
largest_std_dev
```
